```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("student_score.csv")
print(df.head())
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

# **Drop Unnamed Column**

```python
df = df.drop("Unnamed: 0", axis=1)
```

```python
print(df.head())
```

# **Change Weekly Study Hours Column**

```python
df["WklyStudyHours"] = df["WklyStudyHours"].str.replace("05-oct", "5-10")
df.head()
```

# **Gender Distribution**

```python
plt.figure(figsize= (5,5))
ax = sns.countplot(data = df, x = "Gender")
plt.title("Gender Distribution")
ax.bar_label(ax.containers[0])
plt.show()
```

```python
#from the above chart we have analysed that:
#the number of femals in the data is more than the number of males 
```

```python
gb = df.groupby("ParentEduc").agg({"MathScore":"mean","ReadingScore":"mean","WritingScore":"mean"})
print(gb)
```

```python
plt.figure(figsize= (5,5))
plt.title("Relationship between parent's Education and Student's score")
sns.heatmap(gb)
plt.show()
```

```python
#from the above chart we have concluded that the education of the parents have a good impact 
```

```python
gb1 = df.groupby("ParentMaritalStatus").agg({"MathScore":"mean","ReadingScore":"mean","WritingScore":"mean"})
print(gb1)
```

```python
plt.figure(figsize= (3,3))
plt.title("Relationship between parent's marital status and Student's score")
sns.heatmap(gb1)
plt.show()
```

```python
#from the above chart we have concluded that there is no/negligible impact on the 
#student's score due to their parent's marital status
```

# **Distribution of Ethnic Group**

```python
groupA = df.loc[(df["EthnicGroup"] == "group A")].count()
groupB = df.loc[(df["EthnicGroup"] == "group B")].count()
groupC = df.loc[(df["EthnicGroup"] == "group C")].count()
groupD = df.loc[(df["EthnicGroup"] == "group D")].count()
groupE = df.loc[(df["EthnicGroup"] == "group E")].count()
print(groupA["EthnicGroup"])
```

```python
l = ["group A", "group B", "group C", "group D", "group E"]

mlist = [
    groupA["EthnicGroup"],
    groupB["EthnicGroup"],
    groupC["EthnicGroup"],
    groupD["EthnicGroup"],
    groupE["EthnicGroup"]
]

plt.pie(mlist, labels=l, autopct = "%1.2f%%")
plt.title("Distribution of Groups")
plt.show()
```
